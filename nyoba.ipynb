```python
import pandas as pd

# 1. Baca file
df = pd.read_csv('data/raw/cnbc-with-published.csv')

# 2. Ambil timezone unik, buang data yang kosong (dropna), dan ubah jadi list biasa
unique_timezones = df['published_timezone'].dropna().unique().tolist()

print("Timezone yang ada di data kamu:")
for tz in unique_timezones:
    print(f"- {tz}")
```
